### 📌 Análisis individual [3.0 puntos]

#### 📌 Batch 1 (01/01/25 - 05/01/25) [0.75 puntos]

````python
# código para evaluar el performance de su modelo
````

````python
import pandas as pd

def evaluar_performance(predicciones, validacion, numero_total_esperado):
    # Tuplas customer_id - product_id existentes en validación
    tuplas_validacion = set(zip(validacion['customer_id'], validacion['product_id']))
    # Tuplas customer_id - product_id predichas
    tuplas_predichas = set(zip(predicciones[0], predicciones[1]))
    # Verdaderos positivos
    verdaderos_positivos = len(tuplas_validacion.intersection(tuplas_predichas))
    # False positives
    false_positives = len(tuplas_predichas - tuplas_validacion)
    # Verdaderos negativos
    verdaderos_negativos = numero_total_esperado - len(tuplas_validacion) - false_positives
    # False negatives
    false_negatives = len(tuplas_validacion) - verdaderos_positivos

    # F1 Score
    precision = verdaderos_positivos / (verdaderos_positivos + false_positives)
    recall = verdaderos_positivos / (verdaderos_positivos + false_negatives)
    f1_score = 2 * (precision * recall) / (precision + recall)

    return f1_score, precision, recall


# cargar clientes
clientes = pd.read_parquet('data/clientes.parquet')
# número de clients
num_clientes = clientes['customer_id'].nunique()

# cargar productos
productos = pd.read_parquet('data/productos.parquet')
# número de productos
num_productos = productos['product_id'].nunique()

# Total de predicciones esperadas
total_predicciones_esperadas = num_clientes * num_productos
````

````python
# cargar las predicciones
predicciones_1 = pd.read_csv('predictions/semana_1.csv', header=None)

# cargar batch de datos de validación
validacion_1 = pd.read_parquet('data/batch_t1.parquet')

f1, precision, recall = evaluar_performance(predicciones_1, validacion_1, total_predicciones_esperadas)
print(f'F1 Score: {f1:.4f}, Precision: {precision:.4f}, Recall: {recall:.4f}')
print(f'Número total de predicciones esperadas: {total_predicciones_esperadas}')
print(f'Número de compras reales en validación: {len(validacion_1)}')
````

El recall del modelo indica que se está identificando un 80% de las compras reales realizadas por los clientes en este período. Aunque esto viene a costa de una precisión más baja (24%), lo que sugiere que muchas de las predicciones positivas son falsas alarmas. Pero como no se desean perder posibles ventas, este trade-off puede ser aceptable en el contexto del negocio.

#### 📌 Batch 2 (06/01/25 - 12/01/25) [0.75 puntos]

````python
# código para evaluar el performance de su modelo

# carga de predicciones y datos de validación para semanas siguientes
predicciones_2 = pd.read_csv('predictions/semana_2.csv', header=None)
validacion_2 = pd.read_parquet('data/batch_t2.parquet')

f1_2, precision_2, recall_2 = evaluar_performance(predicciones_2, validacion_2, total_predicciones_esperadas)
print(f'F1 Score Semana 2: {f1_2:.4f}, Precision: {precision_2:.4f}, Recall: {recall_2:.4f}')
print(f'Número de compras reales en validación: {len(validacion_2)}')
````

Esta segunda semana se observa una ligera disminución en el recall (77%) y un aumento en la precisión (35%). Lo que conlleva un aumento importante en el F1 Score a 48%. Dado que los batches son de una sola semana, es poco probable que la semana extra de entreno haya tenido un impacto significativo en el modelo. Sin embargo es probable que las predicciones de está semana hayan sido más faciles, ya que la semana anterior corresponde a los días de comienzo de año nuevo, donde los patrones de compra pueden ser más erráticos. Cabe mencionar que como el periodo de entrenamiento corresponde a aprox un año, es poco probable que el modelo haya logrado captar patrones anuales.

````python
# código para evaluar el performance de su modelo

# carga de predicciones y datos de validación para semanas siguientes
predicciones_3 = pd.read_csv('predictions/semana_3.csv', header=None)
validacion_3 = pd.read_parquet('data/batch_t3.parquet')
f1_3, precision_3, recall_3 = evaluar_performance(predicciones_3, validacion_3, total_predicciones_esperadas)
print(f'F1 Score Semana 3: {f1_3:.4f}, Precision: {precision_3:.4f}, Recall: {recall_3:.4f}')
print(f'Número de compras reales en validación: {len(validacion_3)}')
````
